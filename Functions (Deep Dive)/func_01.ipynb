```python
# without decorators

def calc_Total():
    print("total price is 100")

calc_Total()    
```

```python
# with decorators

def my_decorators(func):
    def wrapper():
        print("before function")
        func()
        print("after function")

    return wrapper

def greet():
    print("hello")

greet1 = my_decorators(greet)

greet1()
```

```python
# with @ decorator

def decorator(func):
    def wrapper():
        print("starting")
        func()
        print("ending")

    return wrapper

@decorator
def msg():
    print("heyyyy")

msg()            
```

```python
def myfunction(func):
    def wrapper():
        print("before the function is called")
        func()
        print('after the functin is called')

    return wrapper

@myfunction
def say_hello():
    print("hello")

say_hello()        
```

```python
def myfunction(func):
    def wrapper(a,b):
        print("before the function is called")

        if b>a:
            b,a =a,b
        else:
            b,a = b,a

        print("operation performed")
        func(a,b)
        print("after the function is called")

    return wrapper

@myfunction
def dividenumber(a,b):
    c = a/b
    print(c)

dividenumber(20,60)   
dividenumber(60,10)             
```

```python
def calculations(func):
    def wrapper(a,b):
        print("calculations started")
        func(a,b)
        print("calculation completed")

    return wrapper


@calculations
def add_num(a,b):
    print("sum",a + b)
    

add_num(45,56)        
```

```python
def positive_only(func):
    def wrapper(number):
        if number >0:
            return func(number)

        else:
            return "pls enter a positive number"

    return wrapper

@positive_only
def square(number):
    return number**2

print(square(5))
print(square(-3))
print(square(0))    

```

```python
def int_only(func):
    def wrapper(number):
        if isinstance(number,int):
            return func(number)
        else:
            return "error: pls enter integer"

    return wrapper

@int_only
def double(number):
    return number * 2       

print(double(10))
print(double("10"))
print(double(3.5))    
```

```python

def calculation(func):
    def wrapper(x,y):
        if isinstance(x,str):
            x = int(x)
        if isinstance(y,str):
            y = int(y)

        return func(x,y)

    return wrapper            


@calculation
def division(x,y):
    print("division:",x/y)

division(10,2)
division("10",5)
 
```

```python
def marks_check(func):
    def wrapper(marks):
        if not isinstance(marks,(int,float)):
            return "marks must be number"

        elif marks<0 or marks>100:
            return "marks must be between 0 to 100"

        else:
            return func(marks)

    return wrapper        

@marks_check
def result(marks):
    if marks >=35:
        return "pass"
    else:
        return "fail"   


print(result(78))
print(result(32))
print(result(110))
```

```python
def  check_quantity(func):
    def wrapper(quantity):
        if isinstance(quantity,bool):
            return "boolean values are not allowed"

        if isinstance(quantity,str):
            if quantity.isdigit():
                quantity = int(quantity)
            else:
                return "String must contain digits only"


        if not isinstance (quantity,int):
            return "quantity must be an integer"   


        if quantity < 1:
            return "quantity must be at least 1"


        return func(quantity)

    return wrapper


@check_quantity
def total_items(quantity):
    return quantity * 100


print(total_items(5))
print(total_items("8"))
print(total_items("abc"))
print(total_items(0))
print(total_items(2.5))
print(total_items(True))


         
```

```python
def clean_number( func):
    def wrapper(number):
        if not isinstance(number,(int,float)):
            print("enter a number")

        if isinstance(number,bool):
            print("boolean is not allowed")

        number = round(abs(number),2)
        return func(number)

    return wrapper

@clean_number
def display_num(number):
    print("amount:",number)


display_num(-25.678)
display_num(12.345)
display_num("25")            
```

```python
# List 
List_01 = [1, 2, 3, 4]

# Decorator 
def my_decorator(func):
    def wrapper(*args, **kwargs):
        print("--- Start ---")
        func(*args, **kwargs)
        print("--- End ---")
    return wrapper

@my_decorator
def originalfunc(List_01):
    for i in List_01:
        print(i**2)

originalfunc(List_01)
```
