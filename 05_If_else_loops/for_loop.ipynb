# For Loop

```python
# range

for i in range(0,10):
    print(i)
```

```python
newDict = {'name':"RAJESH",'AGE':32}

for key,values in newDict.items():
    print(key,values)
```

```python
# Set Datatypes

setOne = {1,2,'q','v','b'}

for i in setOne:
    print(i)
```

```python
# String

name = 'Rajesh'

for i in name:
    print(i)
```

```python
newDict = {'name':'RAJESH','AGE':32}

for i in newDict.items():
    print(i)
```

```python
fruitList = ['apple','banana','cherry','kiwi']

for i in fruitList:
    print(i)
```

```python
num = [1,2,3,4,5,6,7,8]

for i in num:
    print(i)
```

```python
# Tuple

fruits = ('apple','banana','cherry','kiw')

for i in fruits:
    print(i)
```

```python
range1 = range(1,5)

for i in range1:
    print(i)
```

```python
# Nested Loop

# Find the matching element from two lists

List_01 = ['Red','Yellow','Lightgreen']
List_02 = ['White','Black','Yellow']

for i in List_01:
    for j in List_02:
        if i==j:
            print('matched:',i,j,)
```

```python
adj = ['Red','Yellow','Lightgreen']
fruits = ['Apple','Banana','Grapes']

for i in adj:
    for j in fruits:
        print(i,j)
```

```python
firstList = [3,2,1]
secondList = [1,2,3]

for i in firstList:
    for j in secondList:
        if i==j:
            print(i,j)
```

```python
for j in range(1,7):
    for i in range(1,j):
        print('*',end=" ")
        print()
```

```python
for even_num in range(2,21,2):
    print(even_num)
```

```python
total_sum = 0

for i in range(1,11):
    total_sum = i+total_sum
    print(total_sum)
```

```python
sales_numbers = [12,7,18,5,24,9,30]

even_count = 0
odd_count= 0

for values in sales_numbers:
    if values % 2==0:
        even_count +=1
    else:
        odd_count += 1

print(even_count)
print(odd_count)            
```

```python
marks =[35,78,92,45,61,28,85]

for mark in marks:
    if mark>50:
        print(mark)
```

```python
numbers =[2,4,6,45,3,2,9,8,8]

for num in numbers:
    if num>10:
        print(num)
```

```python
word_text ="education"
vowel_count =0

for character in word_text:
    if character in "aeiou":
        vowel_count +=1

print(vowel_count)
```

```python
city_name = "Pune"
reversed_name = ""

for character in city_name:
    reversed_name = character + reversed_name

print(reversed_name)    
```
