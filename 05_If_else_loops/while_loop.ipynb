# While Loop

```python
# print number from 0 to 10

i =0

while i<=10:
    print(i)
    i = i + 1
```

```python
# print only even numbers from 0 to 10

i = 0

while i <= 10:
    if i%2 ==0:
        print(i)

    i = i + 1    
```
