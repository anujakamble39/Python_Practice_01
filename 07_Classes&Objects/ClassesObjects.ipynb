```python
class Employee:
    def __init__(self,name,salary):
        self.name = name
        self.salary = salary
    def calc_bonus(self):
        if self.salary <30000:
            bonus = 5000
        else:
            bonus =20000

        print("Employee:",self.name)
        print("Salary:",self.salary)
        print("Bonus:",bonus)

emp1 = Employee("Anuja",25000)
emp2 = Employee("Ganesh",40000)

emp1.calc_bonus()
emp2.calc_bonus()
```

```python
class Student:
    def __init__(self,name,marks):
        self.name = name
        self.marks = marks 

    def check_result(self):
        if self.marks >=75:
            print(self.name,"Excellent")

        elif self.marks>=50:
            print(self.name,"Good")

        else:
            print(self.name,"needs improvement")

stu1 = Student("anuja",79)
stu2 = Student("ganesh",56)
stu3 = Student("sneha",45)

stu1.check_result()
stu2.check_result()
stu3.check_result()
```

```python
class Student:
    def __init__(self,name,marks):
        self.name = name
        self.marks = marks

stu1 = Student("anuja",79)
stu2 = Student("ganesh",56)
stu3 = Student("sneha",45)  

students=[stu1,stu2,stu3]
total_marks = 0

for student in students:
    total_marks = total_marks + student.marks

print("Total_marks:",total_marks)    
```
