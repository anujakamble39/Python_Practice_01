```python
#loop

values = ["10","20",'abc','40']
total = 0

for value in values:
    try:
        total = total + int(value)
    except ValueError:
        print("SKIPPED:",value)
print("Total:",total)            
```

```python
values = ["5", "abc", "10", "xyz", "20"]
total = 0

for value in values:
    try:
        total += int(value)
    except ValueError:
        continue

print(total)
```

```python
def divide(a, b):
    try:
        return a / b
    except ZeroDivisionError:
        return "Cannot divide by zero."


print(divide(20, 4))
print(divide(20, 0))
```
