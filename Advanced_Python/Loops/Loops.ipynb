# Nested loops

```python
for class_num in range(1,4):
    for stu_num in range(1,3):
        print(class_num,stu_num)
```

```python
for number in range(2,5):
    for multiplier in range(1,6):
        print(number,"x",multiplier,"=",number*multiplier)

print()
```

```python
departments = ["HR","Sales"]
employees = ["Anuja","Ganesh","Sneha"]

for department in departments:
    for employee in employees:
        print(department,"-",employee)
```

```python
col = ["red",'yellow','lightgreen']
fruits = ['apple','banana','grapes']

for i in col:
    for j in fruits:
        print(i,j)
```

```python
col_01 = ["Red", "Yellow", "LightGreen"]
col_02 = ["Black","White" ,"LightGreen"]


for i in col_01:
    for j in col_02:
        if i==j:
            print("Matched",i,j)
```

```python
col_01 = ["Red", "Yellow", "LightGreen"]
col_02 = ["Black","White" ,"LightGreen"]


def func_01(col_01,col_02):
    for i in col_01:
        for j in col_02:
            if i==j:
                return "Matched",i,j

values = func_01(col_01,col_02)
values
type(values)            
```
