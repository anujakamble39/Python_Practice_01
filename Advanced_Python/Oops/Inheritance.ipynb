```python
class Employee:
    def __init__(self,name,salary):
        self.name = name 
        self.salary = salary

    def show_details(self):
        print("Name:",self.name)
        print("Salary:",self.salary)

class Analyst(Employee):
    def analyze_data(self):
        print(self.name, "is analyzing data")

analyst1 = Analyst('anuja',45000)
analyst2 = Analyst("ganesh",56000)

analyst1.show_details()
analyst1.analyze_data()

print("============")

analyst2.show_details()
analyst2.analyze_data()


```

```python
class Product:
    def __init__(self,name,price):
        self.name = name
        self.price = price

    def show_product(self):
            print("product name:",self.name)
            print("Price:",self.price)

class electronicpro(Product):
     def category(self):
          print(self.name, "is electronics")


ele1 = electronicpro("keyboard",1200)

ele1.show_product()
ele1.category()
```

```python
class vehicle:
    def __init__(self,brand):
        self.brand = brand

class car(vehicle):
    def __init__(self,brand,fueltype):
        super().__init__(brand)
        self.fueltype = fueltype

    def display(self):
        print("Brand:",self.brand)
        print("Fueltype:",self.fueltype)

car1 = car('TaTa','petrol')
car1.display()                     
```

# Multilevel inheritance

```python
class GrandParentclass:
    def GP_method(self):
        return "GP method I have 50 acres of land"

class Parentclass(GrandParentclass):
    def p_class(self):
        return "p class I have 50 acres of land givn by gp method"

class childclass(Parentclass):
    def c_class(self):
        return "c class I have 50 acres of land given by p class"

cc1 = childclass()
cc1.GP_method()
cc1.p_class()
cc1.c_class()    
```

```python
class device:
    def show_device(self):
        print("this is device")

class computer(device):
    def show_details(self):
        print("the device is computer")

class laptop(computer):
    def details(self):
        print("this device is laptop")

device1 = laptop()

device1.show_device()
device1.show_details()
device1.details()
```

# Multiple inheritance

```python
class camera:
    def take_phota(self):
        print("capture photos")

class phone:
    def make_call(self):
        print("calling anyone")

class smartphone(camera,phone):
    def show_detail(self):
        print("can search anything")


obj1 = smartphone()
obj1.take_phota()
obj1.make_call()
obj1.show_detail()                        
```

```python
class printer:
    def print_doc(self):
        print("can print all document")

class scanner:
    def scan_doc(self):
        print("can scan all document")

class allinone(printer,scanner):
    def copy_doc(self):
        print("can do both")

obj1 = allinone()
obj1.print_doc()
obj1.scan_doc()
obj1.copy_doc()                        
```

Multilevel inheritance

```python
class Hospital:
    def show_hospital(self):
        print("Hospital: Medicare Hospital")

class Doctor(Hospital):
    def show_doctor(self):
        print("Doctor: Dr.V.S.Deshmukh")

class Surgeon(Doctor):
    def specialization(self):
        print("Specialization: Heart surgery")

s1 = Surgeon()
s1.show_hospital()
s1.show_doctor()
s1.specialization()
                        
```
