# Method Overloading

```python
class myClass:
    def sum_4(self,a,b,c,d):
        return a+b+c+d

    def sum_3(self,a,b,c):
        return a+b+c

    def sum_2(self,a,b):
        return a+b

my_obj = myClass()
my_obj.sum_3(1,2,3)
my_obj.sum_2(1,1)
my_obj.sum_4(1,2,3,4)               
```

```python
class myClass:
    def sum_overload(self,a=None,b=None,c=None,d=None):
        if a!= None and b!= None and c!= None and d!= None:
            return a+b+c+d

        elif a!= None and b!= None c!=None:
            return a+b+c

        elif a!= None and b!= None:
            return a+b

        else:
            return "pls pass atleast two values for sum"

my_obj = myClass()
my_obj = sum_overload(1,2,3,4)            
```

```python
class P_C_Interest:
    def Interest(self):
        return 0.011

class c_c_Home_Interest(P_C_Interest):
    def Interest(self):
        return 0.08

Home_obj = c_c_Home_Interest()
Home_obj.Interest()                
```

```python
# Method OverRiding with super Keyword:

class Animal:
    def speak(self):
        print("The animal makes a sound")


class Dog(Animal):
    def speak(self):
        super().speak()
        print("the dog barks")

dog = Dog()
dog.speak()                
```

```python
class P_C_Interest:
    def Interest(self):
        print(0.011)

class c_c_Home_Interest(P_C_Interest):
    def Interest(self):
        super().Interest()
        print(0.08)

Home_obj = c_c_Home_Interest()
Home_obj.Interest()                
```
