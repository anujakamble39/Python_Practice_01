```python
class temptconverter:
    
    def fahrenheit_c(self,temp):
        self.temp = temp
        return (self.temp *9/5)+32

    def celsius_c(self,temp):
        self.temp = temp
        return (self.temp -32)*5/9

calc_temp = temptconverter()

print(calc_temp.fahrenheit_c(34))
print(calc_temp.celsius_c(23))
```

```python
class Product:
    def __init__(self,name,price,quantity):
        self.name = name
        self.price = price
        self.quantity = quantity

    def display(self):
        print("P_name:",self.name)
        print("P_price:",self.price)
        print("P_quantity:",self.quantity)    

p_01 = Product("laptop",45000,3)
p_02 = Product("mobile",20000,4)

p_01.display()
print("=========")
p_02.display()

print("=========")
print("after")
p_01.quantity = 6
p_02.price = 25000


p_01.display()
print("=========")
p_02.display()






```

```python
class student:

    def __init__(self,name,science,english,math):
        self.name = name
        self.science = science
        self.english = english
        self.math = math

    def display(self):
        print("Name:",self.name)
        print("Science:",self.science)
        print("English:",self.english)
        print("Math:",self.math)
        print("total:",self.science + self.english + self.math)

stu1 = student("anuja",45,67,89)
stu2 = student("ganesh",49,98,67)

stu1.display()
print("-----------")

stu2.display()

stu1.science = 99
stu2.math =100

print("============")

stu1.display()
print("-----------")

stu2.display()


```

```python
class Student:
    def __init__(self,name,science,eng,math):
        self.name = name
        self.science = science
        self.eng = eng
        self.math = math

    def display(self):
        print("Name:",self.name)
        print("Science:",self.science)
        print("Eng:",self.eng)
        print("Math:",self.math)
        print("Total_marks:",self.science+self.eng+self.math)

stu1 = Student("saksham",76,89,56)
stu1.display()
```

```python
class Employee:
    def __init__(self,name,salary):
        self.name = name
        self.salary = salary

    def increment(self,amount):
        self.salary += amount
        

emp1 = Employee("elena",35000)
print("before:",emp1.salary)

emp1.increment(5000)

print("after:",emp1.salary)

```
