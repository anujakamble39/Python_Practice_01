```python
from functools import reduce

number = [1,2,3,4,5,6]

even_num = list(filter(lambda n:n%2==0,number))
print('Even num:',even_num)

square_num = list(map(lambda n:n**2,even_num))
print("Square:",square_num)

total = reduce(lambda a,b: a+b,square_num)
print("Total:",total)
```
