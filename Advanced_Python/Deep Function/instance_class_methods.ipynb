```python
class Dog():
    def __init__(self,dog_age):
        self.dog_age = dog_age

    def abc(self):
        print(self.dog_age)

    def my_dog_details(self,color):
        print(f"my Dog has {color}color")

dog_1 = Dog(14)
dog_1.abc()
dog_1.my_dog_details("brown")                
```

```python
class Dog():

    def __init__(self,dog_age):
        self.dog_age = dog_age

    def display(self):
        print(self.dog_age) 

    def details(self):
        print(f"my dog's age is{self.dog_age}")


dog1 = Dog(14)
dog1.display()
      
```

```python
class Bankig:
    Int_Rate = 0.08     # 8%Class Variable

    @classmethod
    def change_int_Rage(cls):
        cls.Int_Rate = cls.Int_Rate + (cls.Int_Rate*0.02)
        print(cls.Int_Rate)

Bankig.Int_Rate
Bankig.change_int_Rage()
```

```python
class Employee:
    def display_salary(self):
        name = "Rahul"
        monthly_salary = 25000
        annual_salary = monthly_salary * 12

        print("Employee:", name)
        print("Monthly Salary:", monthly_salary)
        print("Annual Salary:", annual_salary)


e1 = Employee()
e1.display_salary()
```
