```python
class Bank:
    def __init__(self,balance):
        self.balance = balance

    def withdraw(self,amount):
        if amount <= self.balance:
            self.balance = self.balance - amount

            print("withdraw successfull")
            print("remaining Balance:",self.balance)

        else:
            print("insufficient balance")


obj1 = Bank(10000)
obj1.withdraw(2000)
obj1.withdraw(4000)
obj1.withdraw(11000)
```

```python
class Employee:
    def __init(self,salary):
        self.__salary =salary
    def check_salary(self):
        if isinstance(self.__salary,(int,float)):
            print("value is integer or float")

            if self.__salary>=25000:
                print("eligible")

            else:
                print("not eligible")

        else:
            print("values must int or float")                

emp1 = Employee(34000)
emp1.check_salary()

emp2 = Employee(23000)
emp2.check_salary()
```
