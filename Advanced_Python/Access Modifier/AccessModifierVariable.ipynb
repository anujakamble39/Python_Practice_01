```python
# Public variable

class Employee:
    def __init__(self,name,salary):
        self.name = name
        self.salary = salary

emp1 = Employee("anu",20000)
print(emp1.name)        
print(emp1.salary)
```

```python
# public ,protected and private varibles

class BankAccount:
    def __init__(self,holder,balance,pin):
        self.holder = holder
        self._balance = balance    # Protected
        self.__pin = pin           # private

    def show_display(self):
        print("Holder:",self.holder)
        print("Balance:",self._balance)
        print("Pin:",self.__pin)

obj1 = BankAccount("anuja",25000,1234)

obj1.show_display()
```

```python
class Product:
    def __init__(self,productname,price,stock):
        self.productname = productname
        self._price = price
        self.__stock = stock

    def display(self):
        print(self.productname)
        print(self._price)
        print(self.__stock)


P01 = Product("laptop",45000,20)

P01.display()
        
```

```python
class Employee:
    def __init__(self,name,dept,salary):
        self.name = name
        self._dept = dept
        self.__salary = salary

    def show_employee(self):
        print(self.name)
        print(self._dept)
        print(self.__salary)

emp1 = Employee("soham","IT",35000)

emp1.show_employee()
```

```python
class Student:
    def __init__(self):
        self.name = "Rahul"   # public
        self._age = 20        # protected
        self.__marks = 90     # private

    def display(self):
        print(self.name)
        print(self._age)
        print(self.__marks)

obj = Student()
obj.display()            
```

```python
class Student:
    def __init__(self):
        self.name = "Rahul"   # public
        self._age = 20        # protected
        self.__marks = 90     # private

obj = Student()

print(obj.name)
print(obj._age)
print(obj.__marks)
```

```python
# inside child class inheritance

class parentclass:
    def __init__(self):
        self.name = "Rahul"          # Public
        self._age = 20               # Protected
        self.__marks = 90            # Private

class childclass(parentclass):
    def display(self):
        print(self.name)
        print(self._age)
        print(self.__marks)

obj1 = childclass()
obj1.display()         
```
