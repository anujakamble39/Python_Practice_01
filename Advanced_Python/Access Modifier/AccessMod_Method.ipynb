```python
# Public protected and private methods

class Calculator:

    def add(self,a,b):    # Public method
        return a + b

    def _multiply(self,a,b):  # protected
        return a * b

    def __subtract(self,a,b):  # private
        return a - b

    

    def calculate(self,x,y):
        print("add:",self.add(x,y))
        print("multiply:",self._multiply(x,y))
        print("subtract:",self.__subtract(x,y))


c1 = Calculator()

print(c1.add(10,5))
print(c1._multiply(10,5))
c1.calculate(10,5)




```

```python
class greeting:
    def public_msg(self):
        print("hello soham")

    def _welcome_msg(self):
        print("welcome soham")

    def __secret_msg(self):
        print("how are you soham")


    def show_all(self):
        self.public_msg
        self._welcome_msg
        self.__secret_msg


greet1 = greeting()

greet1.public_msg()
greet1._welcome_msg()


try:
    greet1.__secret_msg()

except AttributeError as e:
    print("error:",e)   

greet1.__secret_msg()     


```

```python
class Student:

    def display_public(self):
        print("Public Method")

    def _display_protected(self):
        print("Protected Method")

    def __display_private(self):
        print("Private Method")

obj = Student()

obj.display_public()        # Allowed
obj._display_protected()    # Allowed
obj.__display_private()     # Error
```

```python
class Student:

    def display_public(self):
        print("Public Method")

    def _display_protected(self):
        print("Protected Method")

    def __display_private(self):
        print("Private Method")


class Teacher(Student):
    def show(self):
        self.display_public()       # Allowed
        self._display_protected()   # Allowed
        self.__display_private()    # Error


obj = Teacher()
obj.show()
```
