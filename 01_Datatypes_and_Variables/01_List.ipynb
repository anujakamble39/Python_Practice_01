# LIST

```python
# SLICING / ACCESSING

fruits = ["apple","banana","mango"]

print(fruits[0])
print(fruits[-1])
```

```python
cities = ["pune","nashik","mumabai","nagpur"]

print(cities[0])
print(cities[2])
```

```python
colours = ["red","blue","green","yellow"]

print(colours[-1])
print(colours[-3])
```

```python
# change an element

products = ["Laptop","Mouse","Keyboard"]
print(products)

products[1]="Monitor"

print(products)
```

```python
scores = [45,67,72,81]
print(scores)

scores[0]=56
print(scores)
```

```python
# Adding Elements
#append

employees = ["anuja","ganesh","samyak"]
print(employees)

employees.append("nikita")

print(employees)
```

```python
numbers = [11,39,36,56]
print(numbers)

numbers.append(45)
print(numbers)
```

```python
#insert

animals = ["dog","cat","horse"]
print(animals)

animals.insert(2,"sheep")
print(animals)
```

```python
#extends

first_batch = [101,102,103]
second_batch = [104,105]

first_batch.extend(second_batch)

print(first_batch)
```

```python
list_a =[1,2,3]
list_a.extend([4,5])

print(list_a)
```

```python
#Removing Elements
#Remove

language = ["python","java","excel","ruby"]
print(language)

language.remove("ruby")

print(language)
```

```python
#pop

price = [120,250,180,300]
print(price)

removed_price = price.pop(1)
print(price)
print("Removed_price:",removed_price)
```

```python
#del

vehicles = ['car','bike','bus','train']

print(vehicles)

del vehicles[2]
print(vehicles)
```

```python
#clear

task =['study','practice','revision']

task.clear()

print(task)
```

```python
#len

dept = ['HR','sales','finance','IT']
print(dept)

print(len(dept))
```

```python
#count

attendance =['present','absent','present','absent','present']

print(attendance.count('present'))
```

```python
#index

tools = ['python','excel','sql','powerbi']
print(tools)

print(tools[2])
```

```python
#sorting

num = [4,5,3,7,9]
num.sort()
print(num)
```

```python
num.sort(reverse = True)
print(num)
```

```python
#reversing

name = ['anuja','ganesh','saksham','nikita']
print(name)

name.reverse()
print(name)
```

```python
weekdays = ["Monday", "Tuesday", "Wednesday", "Thursday", "Friday"]

print(weekdays[1:4])
```

```python
digits = [0, 1, 2, 3, 4, 5, 6]

print(digits[::2])
```

```python
skills = ["Excel", "SQL", "Power BI"]

print("SQL" in skills)
print("Java" not in skills)
```

```python
batch_01 = ["A","B"]
batch_02 = ["C","D"]

all_batch = batch_01 + batch_02

print(all_batch)
```

```python
marks = [56,39]
print(marks*3)
```

```python
#Nested list

company_data = [
    ["Asha", 45000],
    ["Meena", 52000],
    ["Vijay", 48000]
]

print(company_data[1])
print(company_data[2][1])
```
