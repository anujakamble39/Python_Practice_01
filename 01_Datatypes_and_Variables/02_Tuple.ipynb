# TUPLE

```python
#accessing

states = ("Maharashtra", "Gujarat", "Goa", "Karnataka")
print(states)

print(states[0])
print(states[2])
```

```python
#Negative indexing

planets = ("Earth", "Mars", "Jupiter", "Saturn")
print(planets)

print(planets[-1])
print(planets[-2])
```

```python
#length

vegetables = ("Carrot", "Potato", "Tomato", "Onion")
print(vegetables)

print(len(vegetables))
```

```python
temperatures = (28, 31, 29, 33, 27)

print(temperatures[1])
print(temperatures[4])
```

```python
#Slicing Tuple

regions = ("North", "South", "East", "West", "Central")

print(regions[1:4])
```

```python
ratings = (2,4,5,3,1,5,4)

print(ratings[2:6])
```

```python
departments = ("HR", "IT", "Sales", "Finance", "Marketing")

print(departments[:3])
```

```python
#Negative Slicing

numbers = (23,53,76,78,54,)

print(numbers[-3:])
```

```python
#Modify tuple

food_items =('rice','roti','roti')
print(food_items)

temporary_list = list(food_items)

temporary_list[1]="egg"

food_items = tuple(temporary_list)

print("Modify_items:",food_items)
```

```python
marks = (56,39,45,11)
print(marks)

update_marks = list(marks)
update_marks.append(69)

marks = tuple(update_marks)

print(marks)
```

```python
#Tuple concatenation

old_batch = ('A','B','C')
new_batch = ('D','E')

all_batch = old_batch + new_batch

print(all_batch)
```

```python
#Count

attendance = ("Present", "Absent", "Present", "Present", "Absent")

print(attendance.count("Present"))
```

```python
fruits = ('apple','banana','apple','mango','apple')

print(fruits.count("apple"))
print(fruits.count("banana"))
```

```python
#Index

subjects = ("Maths", "Science", "English", "History")

print(subjects.index("English"))
```

```python
#loop in tuple

countries = ("India", "Japan", "Canada", "Brazil")

for country in countries:
    print(country)
```

```python
numbers = (450, 780, 320, 900)

for number in numbers:
    print(number)
```

```python
job_roles = ('data analyst','mis executive','business analyst')

if "data analyst" in job_roles:
     print('job role is available')
```

```python
# MIN , MAX ,SUM
test_scores = (67, 89, 72, 94, 81)

print("Minimum:", min(test_scores))
print("Maximum:", max(test_scores))
print("Total:", sum(test_scores))
```

```python
#Sorting

numbers = (45, 12, 78, 23, 56)

number_2 = sorted(numbers)

print(number_2)

```

```python
#Reverse

city = ('pune','nagpur','pusad','pandharpur')
reverse_city = city[::-1]

print(reverse_city)
```

```python
#Nested tuple

employee_data = (
    ("Neha", 45000),
    ("Rohan", 52000),
    ("Priya", 48000)
)

print(employee_data)

print(employee_data[0])
print(employee_data[1][1])
```

```python
# tuple values assigned to separate variables

customer = ("anuja",23,"pusad")

name,age,city = customer

print(name)
print(age)
print(city)
```

```python
sales_data = (
    ("January", 125000),
    ("February", 138000),
    ("March", 152000),
    ("April", 141000)
)

for month, salary in sales_data:
    if salary > 140000:
        print(month, salary)
```
