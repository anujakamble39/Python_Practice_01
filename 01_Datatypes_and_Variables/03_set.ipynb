# SET

```python
fruits = {'apple','banana','mango'}
print(fruits)
```

```python
#using set

cities = {'pune','nashik','mumbai','pune'}
city_set = set(cities)

print(city_set)
```

```python
#add

skills = {'excel','sql','powerbi'}
skills.add('python')

print(skills)
```

```python
#Update

numbers = {5,10,15}
numbers.update({20,25,30})

print(numbers)
```

```python
#Remove

language = {'python','java','sql','R'}
print(language)

language.remove("R")
print(language)
```

```python
#Discard

marks = {45,32,76,34}
marks.discard(45)
print(marks)
```

```python
#Pop

colour = {'red','pink','blue','orange'}
removed_colour = colour.pop()

print("Removed:",removed_colour)
print("Remaining:",colour)
```

```python
#Clear

emp_id = {23,45,67,98}
emp_id.clear()
print(emp_id)
```

```python
#Len

dept = {'HR','Sales',"IT",'Finance'}
print(len(dept))
```

```python
#in and not in

skills = {'excel','sql','powerbi'}

print('sql'in skills)
print('java' in skills)

print("sql"not in skills)
print("java" not in skills)
```

```python
#UNION

batch_a ={'A','B','C'}
batch_b ={'D','E','F'}

all_batch = batch_a | batch_b

print(all_batch)
```

```python
#UNION

batch_a ={'A','B','C'}
batch_b ={'D','E','F'}

all_batch = batch_a.union(batch_b)

print(all_batch)
```

```python
# Intersection 

excel_learners = {"Anuja", "Ganesh", "Pooja", "Nikita"}

sql_learners = {"Pooja", "Nikita", "Aman", "Sneha"}

common_learners = excel_learners.intersection(sql_learners)

print(common_learners)
```

```python
#Difference

team_a = {'anuja','ganesh','nikita','akanksha'}
team_b = {'nikita','akanKsha','swara'}

only_team_a = team_a.difference(team_b)

print(only_team_a)
```

```python
# Symmetric Difference

morning_batch = {"Aarav", "Isha", "Kabir", "Tara"}

evening_batch = {"Kabir", "Tara", "Nisha", "Dev"}

different_students = morning_batch.symmetric_difference(evening_batch)

print(different_students)
```

```python
#copy

skills = {"Excel", "SQL", "Python"}

new_skills = skills.copy()

new_skills.add('power BI')

print(new_skills)
```
