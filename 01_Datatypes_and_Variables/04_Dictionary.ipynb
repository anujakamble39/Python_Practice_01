# Dictionary

```python
#create dict

dict_01 = {"anuja","ganesh","sakshi"}

print(dict_01)
```

```python
# key access value

employee = {
    "name": "Sneha",
    "department": "Sales",
    "salary": 35000
}

print(employee["name"])
```

```python
#get 

customer = {
    "name": "Amit",
    "city": "Mumbai",
    "age": 28
}

print(customer.get("city"))
```

```python
#add

student_info = {
    "name": "Neha",
    "age": 21
}

student_info["course"] = "Data Analytics"

print(student_info)
```

```python
#update

student_info = {
    "name": "Neha",
    "age": 21
}

print(student_info)

student_info["age"] = 23

print(student_info)
```

```python
#pop

order ={
    "order_id": 501,
    "customer": "Riya",
    "amount": 2400
}
removed = order.pop('amount')
print(order)
print(removed)
```

```python
employee_data = {
    "name": "Kiran",
    "role": "Analyst",
    "location": "Pune"
}

print(employee_data.keys())
```

```python
#values

employee_data = {
    "name": "Kiran",
    "role": "Analyst",
    "location": "Pune"
}

print(employee_data.values())
```

```python
sales_data = {
    "Laptop": 75000,
    "Mouse": 2500,
    "Keyboard": 4000
}

for product_name,product_price in sales_data.items():
    print(product_name,product_price)
```

```python
#copy

original_data = {
    "name": "Pooja",
    "age": 24,
    "city": "Nagpur"
}

new_data = original_data.copy()

new_data ['course']='data analytics'

print(new_data)
```

```python
#setdefault

original_data = {
    "name": "Pooja",
    "age": 24,
    "city": "Nagpur"
}

original_data.setdefault('course','data analytics')

print(original_data)
```
