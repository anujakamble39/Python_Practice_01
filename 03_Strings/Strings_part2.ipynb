# STRING INDEXING AND SLICING

```python
# POSITIVE INDEXING

language = "Python"

print(language[0])
print(language[1])
print(language[2])
```

```python
course = "Analytics"

print(course[0])
print(course[4])
print(course[8])
```

```python
# NEGATIVE INDEXING

word = "Python"

print(word[-1])
print(word[-2])
print(word[-3])
```

```python
name = "Anuja"

print(name[2])
print(name[-2])
```

```python
word = "Data"

print(word[0])
print(word[-1])
```

```python
# INDEXING WITH VARIABLES

language = "Python"

position = 3

print(language[position])
```

```python
course = "Analytics"

position = -1

print(course[position])
```

```python
# INDEXERROR

word = "Python"
print(word[10])
```

```python
word = "python"
print(len(word)-1)
```

**String Slicing**

```python
word = 'Python'
print(word[0:3])
```

```python
city = "Mumbai"
print(city[0:5])
```

```python
state = "Maharashtra"

print(state[3:6])
```

```python
country = "Yavatmal"

print(country[4:8])
```

```python
word = "Mylife"

print(word[:3])
```

```python
items = "Laptop"
print(items[3:])
```

```python
name = "Anuja"

print(name[:])
```

```python
#SLICING WITH NEGATIVE INDEX

word = "Python"

print(word[-4:-1])
```

```python
course = "Analytics"

print(course[-5:-1])
```

```python
# SLICING WITH STEP

state = "Maharashtra"

print(state[2:6:2])
```

```python
numbers = "12345678"
print(numbers[2:4:2])
```

```python
word = "PythonProgramming"

print(word[0:14:3])
```

```python
fruit = "Pineapple"

print(word[::-1])
```

```python
word = "Earth"
print(word[5:1:-1])
```

```python
name = "Anuja"
print(name)

reversed_name = name[::-1]
print(reversed_name)
```

```python
full_name = "Anuja Kamble"

first_name = full_name[:5]
print(first_name)
```
