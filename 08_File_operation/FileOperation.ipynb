```python
# word count

with open("course.txt", "w") as file:
    file.write("Python is easy.\n")
    file.write("I am learning Python.\n")
    file.write("Python is useful.")

with open("course.txt", "r") as file:
    data = file.read()

count = data.count("Python")

print("Python appears:", count, "times")
```
