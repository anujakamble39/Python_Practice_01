# Conditional filtering

```python
import pandas as pd

file_Path = "G:/Ethans DS & DA/Ethans-main/Data/Airline/Small Airline.csv"
df = pd.read_csv(file_Path)
df
```

```python
cond_01 = df['Age']>=42
cond_02 = df['Airline']=="Jet Airways"
cond_03 = df['Full Name'].str.contains('Gurjar')

#return full dataframe with condition
df[cond_01&cond_02&cond_03]
```

```python
#use of loc
df.loc[cond_01&cond_02&cond_03,['Age','Airline',"Full Name"]]
```

```python
#use of iloc
df.iloc[cond_01&cond_02&cond_03,[1,2,3.4]]
```

```python
#use of iloc
df.iloc[cond_01&cond_02&cond_03,2:5]
```

```python
cond01 = df["Last_name"]=='Kumar'
df[cond01]
```

```python
cond01 = df["Last_name"]=='Kumar'
cond02 = df['City']=="Pune"
cond03 = df['Airline']=='Jet Airways'

df[cond01&cond02&cond03]
```

```python
cond01 = df['Age']>=40
cond02 = df['Airline']=='Multiple carriers'
df[cond01 & cond02]
```
