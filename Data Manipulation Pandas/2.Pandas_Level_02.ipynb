```python
import pandas as pd

data = {
    'Name': ['Anuja', 'Priya', 'Rahul', 'Sneha', 'Amit'],
    'Age': [23, 25, 22, 24, 27],
    'Department': ['HR', 'IT', 'Sales', 'IT', 'HR'],
    'Salary': [25000, 40000, 20000, 35000, 45000]
}

df = pd.DataFrame(data)

print(df)
```

```python

result = df[df['Salary']>30000]

print(result)
```

```python
result = df[(df['Salary']>30000)&(df['Age']>24)]

print(result)
```

```python
result = df[(df['Salary']>30000)|(df['Age']>24)]

print(result)
```

```python
result = df[~(df['Salary']>30000)&(df['Age']>24)]

print(result)
```

```python
result = df[(df['Salary']==30000)&(df['Age']>24)]

print(result)
```

```python
import pandas as pd
result = df[df['Department'].isin(['HR',"IT"])]

print(result)
```

# Filtering using str.contains

```python
# using loc 

result = df.loc[
    df['Salary']>30000,
    ['Name','Salary']
]

print(result)
```

# Concatenation in Pandas

```python
import pandas as pd

df1 = pd.DataFrame({
    'Name': ['Anuja', 'Priya'],
    'Age': [23, 25]
})

df2 = pd.DataFrame({
    'Name': ['Rahul', 'Sneha'],
    'Age': [22, 24]
})

```

```python
print(df1)
print(df2)
```

```python
result = pd.concat([df1,df2])

print(result)
```

```python
result = pd.concat(
    [df1,df2],
    ignore_index=True)

print(result)
```

```python
df1 = pd.DataFrame({
    'Name': ['Anuja', 'Priya'],
    'Age': [23, 25]
})

df2 = pd.DataFrame({
    'Salary': [25000, 40000],
    'Department': ['HR', 'IT']
})


```

```python
result = pd.concat([df1,df2],axis=0)
print(result)
```

```python
result = pd.concat([df1,df2],axis=1)
print(result)
```

# Joins

```python
df1 = pd.DataFrame({
    'Name': ['Anuja', 'Priya', 'Rahul'],
    'Age': [23, 25, 22]
})

df2 = pd.DataFrame({
    'Salary': [25000, 40000, 20000],
    'Department': ['HR', 'IT', 'Sales']
})
```

```python
df1
```

```python
df2
```

```python
result = df1.join(df2)

print(result)
```

```python
result = df1.join(df2,how='left')

print(result)
```

```python
result = df1.join(df2,how='right')

print(result)
```

```python
result = df1.join(df2,how='inner')

print(result)
```

```python
import pandas as pd

employees = pd.DataFrame({
    'Emp_ID': [101, 102, 103],
    'Name': ['Anuja', 'Priya', 'Rahul']
})

salary = pd.DataFrame({
    'Emp_ID': [101, 102, 104],
    'Salary': [25000, 40000, 35000]
})
```

```python
result = pd.merge(
    employees,
    salary,
    on='Emp_ID',
    how='inner'
)

print(result)
```

```python
result = pd.merge(
    employees,
    salary,
    on='Emp_ID',
    how='left'
)

print(result)
```

```python
import pandas as pd

employees = pd.DataFrame({
    'Name': ['Anuja', 'Priya', 'Rahul'],
    'Age': [23, 25, 22]
}, index=[101, 102, 103])

print(employees)
```

```python
df1 = pd.DataFrame({
    'Name': ['Anuja', 'Priya', 'Rahul']
}, index=[101, 102, 103])

df2 = pd.DataFrame({
    'Salary': [25000, 40000, 35000]
}, index=[101, 102, 104])
```

```python
result = df1.join(df2, how='left')

print(result)
```

```python
result = pd.merge(
    df1,
    df2,
    left_index=True,
    right_index=True,
    how='inner'
)

print(result)
```
