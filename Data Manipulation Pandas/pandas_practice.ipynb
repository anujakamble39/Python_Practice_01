```python
import pandas as pd
import numpy as np

data = {
    "Name": ["Amit", "Rahul", "Priya", "Sneha"],
    "Age": [23, np.nan, 22, 24],
    "Salary": [25000, 30000, np.nan, 40000]
}

df = pd.DataFrame(data)

print(df)
```

```python
df.isna()
```

```python
df.isnull()
```

```python
df.isna().sum().sum()
```

```python
miss_percentage = (
    df.isna().mean()*100
)

print(miss_percentage)
```

```python
df.dropna()
```

```python
df.dropna(axis=1)
```

# Fill na

```python
new_df = df.fillna(0)

print(new_df)
```

```python
df["Age"]=df["Age"].fillna(
    df["Age"].mean()
)
print(df)
```

```python
df["Age"] = df["Age"].ffill()
print(df)
```

```python
df["Country"]='India'

print(df)
```

```python
result = df.groupby("City")["Sales"].agg(
    ["sum", "mean", "max", "min", "count"]
)

print(result)
```
