```python
# Use of yield Keyword

def newFunc(a):
    for i in range(1,a):
        yield i
values_1 =newFunc(10)

print(next(values_1))
print(next(values_1))
print(next(values_1))
```

```python
#find largest num

def find_largest(a,b,c):
    if a>=b and a>=c:
        return a
    elif b>=a and b>=c:
        return b
    else:
        return c

result = find_largest(20,48,18)

print(result)
```

```python
def calc_marks(maths,science,english):
    total = maths + science + english
    average = total/3

    return total,average

total_marks,average_marks = calc_marks(80,70,90)

print("Total:",total_marks)
print("Average:",average_marks)
```

```python
def find_largest(a,b):
    if a > b:
        return a
    else:
        return b

print(find_largest(12,9))    
```

```python
def factorial(number):
    result = 1

    for i in range(1,number + 1):
        result = result*i

    return result

print(factorial(5))    
```

```python
def calc_total(numbers):
    total = 0

    for number in numbers:
        total = total + number

    return total

print(calc_total([10,20,30,40]))    
```

```python
def get_num():
    for num in range(1,6):
        yield num

for value in get_num():
    print(value)        
```

```python
def get_num():
    for num in range(1,11):
        if num % 2 ==0:
            yield num
        

for value in get_num():
    print(value)        
```

```python
def square():
    for num in range(1,11):
        yield num * num
        

for value in square():
    print(value)        
```
