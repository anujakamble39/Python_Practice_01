```python
# With Arguments
def printSquares(list_01):
    NewDict_12= ""
    NewDict_12 = {}
    for i in list_01:
        NewDict_12[i] = i*i
    print(NewDict_12)
```

```python
list_02 =  [1,2,3]
printSquares(list_02)
```

```python
list_03 = [10,20,30]
printSquares(list_03)
```

```python
def average_marks(mark1, mark2, mark3):
    average = (mark1 + mark2 + mark3) / 3
    return average

result = average_marks(75, 82, 91)
print("Average:", result)
```

```python
def profit_margin(revenue, cost):
    profit = revenue - cost
    margin = (profit / revenue) * 100
    return margin

answer = profit_margin(100000, 65000)
print("Profit Margin:", answer)
```
